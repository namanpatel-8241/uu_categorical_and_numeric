```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from scipy.stats import beta, pareto, uniform, lognorm, norm, expon
```

```python
x = np.linspace(0.001, 10, 1000)

fig, axes = plt.subplots(2, 3, figsize=(15, 8))

# Beta
a_b, b_b = 2, 5
axes[0,0].plot(x[x<=1], beta.cdf(x[x<=1], a_b, b_b))
axes[0,0].set_title(f'Beta(a={a_b}, b={b_b})')

# Pareto
b_p, xm = 3, 1
axes[0,1].plot(x, pareto.cdf(x, b_p, scale=xm))
axes[0,1].set_title(f'Pareto(b={b_p}, xm={xm})')

# Uniform
lo, hi = 2, 7
axes[0,2].plot(x, uniform.cdf(x, loc=lo, scale=hi-lo))
axes[0,2].set_title(f'Uniform(a={lo}, b={hi})')

# Lognormal
sigma, mu = 0.5, 0
axes[1,0].plot(x, lognorm.cdf(x, s=sigma, scale=np.exp(mu)))
axes[1,0].set_title(f'Lognormal(mu={mu}, sigma={sigma})')

# Normal
mu_n, sigma_n = 5, 1.5
axes[1,1].plot(x, norm.cdf(x, loc=mu_n, scale=sigma_n))
axes[1,1].set_title(f'Normal(mu={mu_n}, sigma={sigma_n})')

# Exponential
lam = 1
axes[1,2].plot(x, expon.cdf(x, scale=1/lam))
axes[1,2].set_title(f'Exponential(lambda={lam})')

for ax in axes.flat:
    ax.set_xlabel('x')
    ax.set_ylabel('density')

plt.tight_layout()
plt.show()

#Used AI for code, to generate refernce distriution.
```

```python
#r=500,dt=0.001, r*dt = 0.5
r = 500
dt = 1e-3
n = 5000
p = r*dt

result4 = []
for i in range(n):
    period = 0
    while True:
        x = rng.choice([0,1],p=(1-p,p))
        period+=1
        if x == 1:
            break  
    time = period*dt
    result4.append(time)

sns.ecdfplot(result4)
```

>- **Exponential Distribution**
>- For a high value of r, say r=500, the probability of the process getting terminated is very high and hence the plot jumps big steps and terminates fast in almost 0.012 peiods.
>- As we decrease the value of r to r=1, the probability r*dt decreases and the process goes for a longer time and shows a exponential CDF distribution.
>- If we decrease dt for high value of r, the process takes smaller steps and gives a smooth curve, but the process still terminates very quickly.
>- For a value r=1 and dt=1e-3, we get a smooth curve for a longer duration.
>- As dt approaches 0, the process increases for a very long period.
